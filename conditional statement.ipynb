```python
country=input("Enter your country")
employment=input("Enter employment status")
salary=int(input("Enter your monthly salary"))
creditscore=int(input("Enter your credit score"))
if country.lower()=="india":
    print("indian")
    if employment=="employed":
        print("emlpoyed")
        if salary>=30000:
            print("eligible")
            if creditscore>=650:
                print("expected credit score")
                print("eligible for loan")
            else:
                print("low credit score")
                print("not eligible")
        else:
            print("low salary")
    else:
        print("unemployed")
else:
    print("non indian")
```

```python
state=input("enter your state")
marks=int(input("enter your marks"))
age=int(input("enter your age"))
qualification=input("enter your qualification")
if state.lower()=="karnataka":
    print("eligible")
    if marks>=50:
        print("passout")
        if age>=18:
            print("adult")
            if qualification.lower()=="puc":
                print("valid for admission")
                print("admission successful")
            else:
                print("invalid for addmission")
                print("admission declined")
        else:
            print("underage")
    else:
        print("less marks")
else:
    print("not our state")
```
